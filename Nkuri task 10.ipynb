```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
```

```python
df = pd.read_csv("Naukri Jobs Data - Mar 2023.csv")
df.head()
```

```python
df.tail()
```

```python
Clean the Data

Handle missing values

Clean salary text (convert to numeric LPA)

Clean experience column

Standardize job titlesb
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
# Replace missing company with a placeholder
df['company'] = df['company'].fillna('Unknown Company')

# Salary cleaning
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')
df['salary'] = df['salary'].fillna(0)

# Job description
df['job-description'] = df['job-description'].fillna('No description provided')

# Fill missing skills
skill_cols = ['skill_1','skill_2','skill_3','skill_4',
              'skill_5','skill_6','skill_7','skill_8']

for col in skill_cols:
    df[col] = df[col].fillna("None")

# Posted date
df['posted on'] = df['posted on'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df['starRating'] = df['starRating'].fillna(0)
df['reviewsCount'] = df['reviewsCount'].fillna(0)
```

```python
df.isnull().sum()
```

```python
top_roles = df['title'].value_counts().head(10)
plt.figure(figsize=(20,3))  # Set figure size
top_roles.plot(kind='bar')
plt.title("Top 10 Job Roles")
plt.xlabel("Job Title")
plt.ylabel("Count")
plt.show()
```

# Top Hiring Locations

```python
plt.figure(figsize=(20,3))

df['location'].value_counts().head(10).plot(kind='bar')

plt.title("Top Hiring Cities")
plt.xlabel("City")
plt.ylabel("Job Count")
plt.show()
```

# Salary Analysis

```python
df.groupby("title")["salary"].mean().sort_values(ascending=False).head(10)
```

```python
plt.figure(figsize=(20,3))

df.groupby("title")["salary"].mean().sort_values(ascending=False).head(10).plot(kind='bar')

plt.title("Top 10 Highest Paying Job Roles")
plt.xlabel("Job Title")
plt.ylabel("Average Salary (LPA)")
plt.show()
```

# Experience Requirement Analysis

```python
df['experience'].value_counts().sort_index().plot(kind='bar')
plt.figure(figsize=(10,5))
plt.title("Experience Required")
plt.xlabel("Years of Experience")
plt.ylabel("Count")
plt.show()
```

# Find most common skills:

```python
df.columns
```

```python
skill_columns = ['skill_1', 'skill_2', 'skill_3', 'skill_4',
                 'skill_5', 'skill_6', 'skill_7', 'skill_8']

# Flatten all skill columns into one list
all_skills = []

for col in skill_columns:
    all_skills.extend(df[col].dropna().astype(str).str.strip())
```

# Now get top 20 skills

```python
from collections import Counter

skill_counts = Counter(all_skills).most_common(20)
skill_counts
```

```python
top_skills = dict(skill_counts)

plt.figure(figsize=(20,6))
plt.bar(top_skills.keys(), top_skills.values())
plt.xticks(rotation=45)
plt.title("Top 20 Most Required Skills")
plt.xlabel("Skills")
plt.ylabel("Count")
plt.show()
```
